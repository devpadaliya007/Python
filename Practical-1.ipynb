## 7. WAP to take 4 subject marks from user and calculate total marks and percentage

```python
c=int(input("Enter C mark here :: "))
dbms=int(input("Enter DBMS mark here :: "))
cn=int(input("Enter CN mark here :: "))
foe=int(input("Enter FOE mark here :: "))
total = c+dbms+cn+foe
avg = total/4
print("Total Marks :: ", total)
print("Percentage :: ", avg)
```

## 12. WAP to convert celcius to fahrenheit and fahrenheit to celcius(f=(9*c/5)+32, c=(f-32)5/9)
