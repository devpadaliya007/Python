```python
def calculator():
    try:
        num1 = float(input("Enter first number: "))
        num2 = float(input("Enter second number: "))
        operator = input("Enter operation (+, -, *, /): ")

        if operator == '+':
            print(f"Result: {num1 + num2}")
        elif operator == '-':
            print(f"Result: {num1 - num2}")
        elif operator == '*':
            print(f"Result: {num1 * num2}")
        elif operator == '/':
            if num2 == 0:
                raise ZeroDivisionError("Cannot divide by zero")
            print(f"Result: {num1 / num2}")
        else:
            raise ValueError("Invalid operator")
    except ValueError as e:
        print(f"Error: {e}")
    except ZeroDivisionError as e:
        print(f"Error: {e}")

calculator()
```

```python
class CustomEligibilityError(Exception):
    pass

def check_blood_donation():
    try:
        age = int(input("Enter your age: "))
        weight = float(input("Enter your weight in kg: "))

        if age <= 18 or weight <= 50:
            raise CustomEligibilityError("You are not eligible to donate blood.")
        print("You are eligible to donate blood.")

    except ValueError:
        print("Invalid input! Please enter numerical values.")
    except CustomEligibilityError as e:
        print(f"Error: {e}")

check_blood_donation()
```

```python
class DivisionByZeroError(Exception):
    pass

def divide_numbers():
    try:
        num1 = float(input("Enter first number: "))
        num2 = float(input("Enter second number: "))

        if num2 == 0:
            raise DivisionByZeroError("Custom Error: Cannot divide by zero")
        
        print(f"Result: {num1 / num2}")
    
    except ValueError:
        print("Invalid input! Please enter numerical values.")
    except DivisionByZeroError as e:
        print(f"Error: {e}")

divide_numbers()
```
