# 1 ) wap to create set using list of element by user

```python
n = int(input("Enter size"))
l = []
for i in range(n):
    l.append(int(input("Enter Number")))
print(l)
print(set(l))
print("Size is",len(l))
```

# 2 ) wap to find minimum & maximum element form set

```python
n = int(input("Enter size"))
s= set()
for i in range(n):
    s.add(int (input("Enter Number")))
print("Maximum is :-",max(s),"Minimum is :- ",min(s))
```

# 3 ) wap to remove element from set

```python
n = int(input("Enter size"))
s= set()
for i in range(n):
    s.add(int (input("Enter Number")))
print(s)
x = int(input("Enter Number for remove"))
if x in s:
    s.remove(x)
    print(s)
else:
    print("Not Found")
```

# 4 )  wap to convert given set into tuple and list

```python
n = int(input("Enter size"))
s= set()
for i in range(n):
    s.add(int (input("Enter Number")))
print(s)
print(tuple(s),end=" ")
print(list(s))
```

# 5 ) wap to perform union , intersection , difference & Symmetric difference opertion on two given set

```python
n = int(input("Enter size"))
n1 = int(input("Enter size"))
s= set()
s1= set()
for i in range(n):
    s.add(int (input("Enter Number")))
for i in range(n1):
    s1.add(int (input("Enter Number")))
print(s,end=" ")
print(s1)
print("Union",s|s1)
print("intersection",s&s1)
print("Difference",s-s1)
print("Symmetric differenc",s^s1)
```
