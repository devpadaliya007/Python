# 11.4  WAP to demonstrate use of date time module  

```python
import datetime as dt
date=input("Enter data as[DD-MM-YYYY]:")
date=dt.datetime.strptime(date,"%d-%m-%Y")
print("Date:",date)
print("Formated Date:",date.strftime("%d,%b %Y %H::%M::%S"))
print("Weekday as[0-6]:",date.weekday())
print("Week day as[1-7]:",date.isoweekday())
print("New Date:",date.replace(hour=4,minute=27,second=52,microsecond=436210))
print("Date:",date.date())
print("Day:",date.day)
print("Month:",date.month)
print("Year:",date.year)
print("Second:",date.second)
print("Microsecond:",date.microsecond)
```

# 11.10 WAP to print person age in years and also print how many days remaining for next birthday

```python
import datetime as dt
dob=input("Enter date as[yyyy-mm-dd]:")
dob=dt.datetime.strptime(dob,"%Y-%m-%d")
today=dt.datetime.now()                                                                         
dt2 = today - dob
print("Age:",dt2.days//365)
if(dob.day,dob.month)>(today.day,today.month):
    next_bd=dob.replace(year=today.year)
else:
    next_bd=dob.replace(year=today.year)
print("remaining days",(next_bd-today).days)
        
```
