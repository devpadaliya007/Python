# 1. w.a.p to enter values and reverse the tuple

```python
n=int(input("enter n:"))
l=[]
for i in range(n):
    l.append(int(input("enter element:")))
tup=tuple(l)
print(tup)
print(tup[::-1])

   
```

# 2.w.a.p to remove duplicate values from tuple

```python
n=int(input("enter n:"))
l=[]
for i in range(n):
    l.append(int(input("enter element:")))
tup=tuple(l)
t=[]
for i in l:
    if i  not in t:
        t.append(i)
print(tuple(t))
```

# 6.3 WAP to check if tuple is distinct or not

```python
n=int(input("enter n:"))
l=[]
for i in range(n):
    l.append(int(input("enter element:")))
tup=tuple(l)
t=[]
for i in l:
    if i  not in t:
        t.append(i)
if len(tup)==len(t):
    print("Distinct")
else:
    print(" Not Distinct:")
```
