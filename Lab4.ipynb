## 1.WAP to find length of given string without using length function.

```python
st1=input("Enter a string :")
count=0
for i in st1:
    count+=1
print("Length :",count)
```

## 2.WAP to Remove ith character from given String.

```python
s=input("Enter a string here :")
a=int(input("Enter Index no of character you want to remove"))
final=s[:a]+s[a+1:]
print(final)
```

## 4.WAP to count no of vowels in  a given string.

```python
s=input("Enter a string here : ")
count=0
for i in s:
    if i=='a' or  i=='e' or i=='i' or i=='o' or i=='u' or i=='A' or i=='E' or i=='I' or i=='O' or i=='U':
        count+=1
print(count)
```

```python
s=input("Enter a string here : ")
count=0
for i in s:
    if i.lower() in ['a','e','i','o','u']:
        count+=1
print(count)
```

## 5.WAP to print even length word in a string.

```python
s=input("Enter a string here : ")
word=s.split(" ")
for w in word:
    if len(w)%2==0:
        print(w)
```

## 6.WAP to cheak whether given string is palindrome or not.

```python
a=input("Enter a string here : ")
if a==a[::-1]:
    print("String is palindrome")
else:
    print("String is not palindrome")
```

## 7.WAP to do string slicing from left rotate and right rotate the given string by d element (where d<=n).

```python
a=input("Enter a string here : ")
d=int(input("Enter Value of d : "))
if d<=len(a):
    lr=a[d:]+a[:d]
    rr=a[d+1:]+a[:d+1]
    print("left rotated string :",lr)
    print("right rotated string :",rr)
```

## 8.WAP to reverse the words in given string.

```python
a=input("Enter a string here : ")
word=a.split(" ")
for i in word[::-1]:
    print(i,end=" ")
```
