# 10.3 WAP to read only special characters from file.

```python
f=open('special.txt','r')
data=f.read()
for ch in data:
    if ch.isalnum()==0:
        print(ch,end='')
f.close()
```

# 10.4.WAP to write N line in new line.

```python
f=open('writefile.txt','w')
n=int(input("Enter line number you want to write :: "))
for i in range(n):
    a=input("Write content :: ")
    data=f.write(a)
    f.write("\n")
f.close()
f=open('writefile.txt','r')
data=f.read()
print(data)
f.close()
```

# 10.6 WAP to write 5 student records(roll no,name,department) in the studentdetails.txt file.

```python
f=open('studentdetails.txt','w')
n=int(input("Enter line number you want to write :: "))
for i in range(n):
    rollno=input("write your rollno :: ")
    data=f.write(rollno)
    f.write("\n")
    
    name=input("write your name :: ")
    data=f.write(name)
    f.write("\n")
    
    department=input("write your department :: ")
    data=f.write(department)
    f.write("\n")
f.close()

f=open('studentdetails.txt','r')
data=f.read()
print(data)
f.close()
```

# 10.7 WAP to find longest word in second.txt file.

```python
with open('secondfile.txt') as f:
    lst= " ".join(f.read().split("\n")).split()
    word= " "
    for i in lst:
        if len(word)<len(i):
                word=i
    print(lst)
    print("Longest Word :- ",word)
```

# 10.9 Wap to append the content of studentdetails.txt file by recording the student record from user

```python
with open("./studentdetails.txt", 'a') as file:
    n = int(input("Enter number of students to append: "))
    for i in range(n):
        file.write(f"Roll No: {input('Enter roll no: ')}\n")
        file.write(f"Name: {input('Enter Name: ')}\n")
        file.write(f"Department: {input('Enter Department: ')}\n")
        file.write("..............................\n") 
file.close()
```
