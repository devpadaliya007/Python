# 1)WAP to cretae class name student and initalize attributes like Er. no,Student name,Gender,Department while createing an object :-

```python
class student:
    def __init__(self,erno,name,gender,dep):
        self.erNo=erno
        self.na=name
        self.gen=gender
        self.dep=dep
    
erno=int(input("Enter Er No.:"))
name=input("Enter Name:")
gender=input("Enter Gender(Male/Female/Other):")
dep=input("Enter Department:")
s=student(erno,name,gender,dep)
print("------------------------------")
print(f"Name:{s.na}")
print(f"ErNo.:{s.erNo}")
print(f"Gender:{s.gen}")
print(f"Department:{s.dep}")
```

# 2)create a class name shape create three sub-classes of shape class name circle,Triangle and squar which content ClaculateArea and DisplayArea methodes.WAP to display the area of all there classes

```python
class shape:                                #Shape Class
    def __init__(self):
        print("Shape Created")
    def calArea():
        print("Calculate Area")
    def displayArea(self):
        print("Display Area")
        
class circle(shape):                        #Circle class
    def __init__(self,r):
        self.radius=r
    def calArea(self):
        self.area=3.14*self.radius*self.radius
    def displayArea(self):
        print(f"Area of Circle is {self.area} which Radius is {self.radius}")
        
class triangle(shape):                      #Triangle Class
    def __init__(self,h,b):
        self.height=h
        self.base=b
    def calArea(self):
        self.area=0.5*self.height*self.base
    def displayArea(self):
        print(f"Area of Trangle is {self.area} which height is {self.height} & Width is {self.base}")
        
class square(shape):                        #Square Class
    def __init__(self,s):
        self.size=s
    def calArea(self):
        self.area=0.5*self.size*self.size
    def displayArea(self):
        print(f"Area of Square is {self.area} which size is {self.size}")

r=int(input("Enter Radiues of Circle"))
b=int(input("Enter Base of Triangle"))
h=int(input("Enter Height of Triangle"))
s=int(input("Enter Size of Square"))
c=circle(r)
t=triangle(h,b)
s=square(s)

c.calArea()           #Circle
c.displayArea()

t.calArea()           #Triangle
t.displayArea()

s.calArea()           #Square
s.displayArea()
```

# 3) Create a class bank_account with Account number,username,E-mail,Account type,AccountBalance data member,Also create a methode addAccountDetail and Display_Accounte_Detailes

```python
class BankAccount:
    def addAccountDetail(self):
        self.AcNo=int(input("Enter Account Number:"))
        self.AcUn=input("Enter Username:")
        self.AcEm=input("Enter E-mail:")
        self.AcTy=input("Enter Account Type:")
        self.AcBa=int(input("Enter Balance:"))
        
    def DisplayAccountDetail(self):
        print("---------------------------")
        print(f"Account Number:{self.AcNo}")
        print(f"UserName:{self.AcUn}")
        print(f"E-mail Id:{self.AcEm}")
        print(f"Account Type:{self.AcTy}")
        print(f"Account Balance:{self.AcBa}")
ac=BankAccount()
ac.addAccountDetail()
ac.DisplayAccountDetail()
```
