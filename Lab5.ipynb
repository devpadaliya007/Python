# 5.6 WAP to print all even numbers of list entered by user 

```python
l=[]
n=int(input("Enter the size : "))
for i in range(n):
    l.append(int(input("Enter the value : ")))
   
for val in l:
    if val%2==0:
        print(val,end=" ")
```

# 5.7 WAP to search element in list

```python
l=[]
n=int(input("Enter the size : "))
for i in range(n):
    l.append(int(input("Enter the value : ")))
s=int(input("Enter the value to search : "))
if s in l:
    print("Found")
else:
    print("Not found.")
```

# 5.8 WAP to sort list in ascending order.

```python
l=[]
n=int(input("Enter the size : "))
for i in range(n):
    l.append(int(input("Enter the value : ")))
l.sort()
print("The sorted list is :-")
for i in l:
    print(i)
```

# 5.9 WAP To convert list of character entered by user into a string

```python
l=[]
n=int(input("Enter the size : "))
for i in range(n):
    l.append(input("Enter the value : ")[0])
print("The string is","".join(l))
```

# 5.10 WAP to interchange the list element based on two position entered by user

```python
l=[]
n=int(input("Enter the size : "))
for i in range(n):
    l.append(int(input("Enter the value : ")))
i1=int(input("Enter first index : "))
i2=int(input("Enter first index : "))
l[i1],l[i2]=l[i2],l[i1]
print("The new list is :-",l)
```

# 5.11 WAP to print all odd numbers of a given list

```python
l=[]
n=int(input("Enter the size : "))
for i in range(n):
    l.append(int(input("Enter the value : ")))
print("The odd numbers are:-")
for val in l:
    if val%2!=0:
        print(val,end=" ")
```
