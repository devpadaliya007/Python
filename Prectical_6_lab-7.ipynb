## 1. WAP to take value in tuple and reverse the tuple.

```python
l=[]
n=int(input("Enter the size : "))
for i in range(n):
    l.append(int(input("Enter the value : ")))
t=tuple(l)
print(t[::-1])
```

## 2. WAP to remove duplicate values from tuple.

```python
l=[]
n=int(input("Enter the size : "))
for i in range(n):
    l.append(int(input("Enter the value : ")))
for i in l:
    if l.count(i)>1:
        for j in range(l.count(i)-1):
            l.remove(i)
        
t=tuple(l)
print(t)
```

## 3. WAP to check if tuple is distinct or not

```python
l=[]
n=int(input("Enter the size : "))
for i in range(n):
    l.append(int(input("Enter the value : ")))
t=tuple(l)
for i in t:
    if t.count(i)>1:
        print("Not Distinct")
        break
else:
    print("Distinct")
```

## 4.WAP to find tuple with positive element in list of tuple

```python
nt=int(input("Enter the number of tuples you want to enter : "))
l=[]
for i in range(nt):
    print(f"\nEnter data for tuple-{i+1}")
    n=int(input("Enter the size : "))
    sl=[]
    for j in range(n):
        sl.append(int(input("\nEnter the value : ")))
    l.append(tuple(sl))
print("The tuples with positive values are :- ")
for i in l:
    for j in i:
        if j<0:
            break
    else:
        print(i)
```

## 5. WAP to find tuple that have all elements divisible by k from list of tuple.

```python
nt=int(input("Enter the number of tuples you want to enter : "))
l=[]
for i in range(nt):
    print(f"\nEnter data for tuple-{i+1}")
    n=int(input("Enter the size : "))
    sl=[]
    for j in range(n):
        sl.append(int(input("\nEnter the value : ")))
    l.append(tuple(sl))
k=int(input("Enter the value of k : "))
print(f"The tuples with all values divisible by {k} are :- ")
for i in l:
    for j in i:
        if j%k!=0:
            break
    else:
        print(i)
```
